# fixed structural prototypes

```python
# Configuration file, do not modify.
config = {'class_num': 94, 'input_dim': 200, "latent_dim": 64, 'model_name': 'site_prob.pt'} 
    
template = Structure.from_file("./POSCAR")  # Read template structure
print("Template file:", template.composition)

# Enter elements not to be replaced. Press Enter directly if none.
no_sub_elem = input('Enter elements that do not need to be replaced:')  
if no_sub_elem == "":
    no_sub_elem = []
else:
    no_sub_elem = no_sub_elem.split(" ")
results = get_templates_from_struc(template, config, no_sub_elem)

# Select top K structures
top_k = 200
results = results[:top_k] 

# Relax structures using general atomic potential function (may take a long time)
# relax_list = []
# for struct in results['structure']:
#     final_structure, final_energy = relax_structure(struct)
#     relax_list.append(final_structure)
# results['relax_structure'] = relax_list

write_results(results, 'ZnS')  # Output the structures to the output folder
```

# fixed element sets

```python
# Configuration file, do not modify.
config = {'class_num': 94, 'input_dim': 200, "latent_dim": 64, 'model_name': 'site_prob.pt'} 

data = pd.read_csv('./data/template.csv')  # Obtain all structural prototypes

# Screen possible compositions via SMACT; element set: ['C', 'B'], 
# maximum atomic number of each element is 5
formula_list = get_filted_composition(['C', 'B'], 3)

for formula in formula_list:
    struc_list= get_templates_from_composition(formula, data)
    results = get_substitue_prob(struc_list, config)

    # Select top K structures
    top_k = 200
    results = results[:top_k] 

    # Relax structures using general atomic potential function (may take a long time)
    # relax_list = []
    # for struct in results['structure']:
    #     final_structure, final_energy = relax_structure(struct)
    #     relax_list.append(final_structure)
    # results['relax_structure'] = relax_list

    write_results(results, formula) # Output the structures to the output folder
```
